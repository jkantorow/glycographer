Scaffold of basic ensemble -> mapper -> volmap usage

```python
import os
import glob

from glycographer.dock import GlycanDockEnsemble
from glycographer.map import VolMap, IntEngMinMapper
```

```python
pose_file_list = glob.glob('test_docked_sample_*')
print(pose_file_list)
```

```python
ensemble = GlycanDockEnsemble.from_poses(
    pose_file_list,
    run_id='test',
    in_complex_pdb='test_complex.pdb',
    grid_pdb='test_receptor_grid.pdb',
    lig_iupac='a-D-Manp-(1->4)-a-L-Fucp'    
)

ensemble.get_score_values()
```

```python
ensemble.to_pdb('test_ensemble.pdb')
```

```python
mapper = IntEngMinMapper(ensemble)
test_map = mapper.map()
```

```python
test_map.ensemble.run_id
```

```python
test_map.to_dx('test_map.dx')
test_map.to_json('test_map_metadata.json')
```

```python
test_map_from_dx = VolMap.from_dx('test_map.dx')
```
